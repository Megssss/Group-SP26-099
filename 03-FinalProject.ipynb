## Research Question

## Background and Prior Work

In 2025, the gaming industry reached a record-breaking $190 billion in market size. Gaming is widespread across the world, with players spanning multiple countries and continents, demonstrating that video games are a shared but culturally influenced activity. Among all genres, action games consistently rank as one of the most popular worldwide, making them a useful focus for analyzing player preferences and consumer behavior.

Previous research has explored how review scores influence video game sales performance. A 2024 study found that higher user review scores were associated with a decrease in total video game sales, while higher critic review scores showed an exponential increase in sales.<a name="ref1"></a><sup>1</sup> These findings suggest that professional critical reception may have a stronger impact on commercial success than player reception alone, highlighting the importance of critic influence and reputation within the gaming market.

A 2021 study found that action games consistently generate the highest global sales revenues across all genres, while North America leads the world in consumer spending on video games.<a name="ref1"></a><sup>2</sup> These trends reinforce the importance of studying action games specifically, as they dominate the market and attract broad international audiences. 

Most recently, a 2026 study found that certain publishers, most notably Nintendo, consistently produce higher-selling games. This suggests that publisher size, name-recognition and reputation contribute heavily to each title's commercial success.<a name="ref1"></a><sup>3</sup> However, the study concluded that no single factor alone can fully explain a game’s performance, emphasizing the need to analyze multiple attributes together.

Building on this prior work, we aim to examine how factors such as Metacritic critic scores, Metacritic user scores, publisher portfolio size, and multi-platform releases relate to and predict total sales for action games released between 2001 and 2020. Metacritic started in 2001 so we will not include any titles released before then. We chose to focus on action games because they consistently dominate global sales revenues and provide a strong basis for analyzing how critical reception, publisher influence, and platform availability contribute to commercial success. Additionally, we seek to determine whether these same relationships can be used to predict total sales for adventure games, allowing us to understand market differences between action and adventure genres.

The findings highlight the importance of examining how review scores, publisher characteristics, and platform availability interact with one another to better understand what contributes to higher sales performance.

### EDA Part 1: Exploring Data Distributions, Anomalies, Skewness, Etc.

#### Distribution of Total Sales

```python
games = pd.read_csv("data/02-processed/game_ultimate_data.csv")
games["log_total_sales"] = np.log10(games["total_sales"] + 1)
games["log_total_sales_per_title"] = np.log10(games["total_sales_per_title"] + 1)
games[["total_sales", "log_total_sales", "total_sales_per_title", "log_total_sales_per_title"]].head()

sns.set_theme(style="whitegrid", context="notebook")
fig, axes = plt.subplots(1, 2, figsize=(18, 6))

# Raw total sales distribution
sns.histplot(data=games, x="total_sales", bins=40, ax=axes[0])
axes[0].set_title("Distribution of Platform-Level Total Sales", fontsize=14, pad=12)
axes[0].set_xlabel("Total Sales", fontsize=12)
axes[0].set_ylabel("Number of Game-Console Releases", fontsize=12)
axes[0].tick_params(axis="both", labelsize=10)

# Log-transformed total sales distribution
sns.histplot(data=games, x="log_total_sales", bins=40, kde=True, ax=axes[1])
axes[1].set_title("Distribution of Log-Transformed Platform-Level Sales", fontsize=14, pad=12)
axes[1].set_xlabel("Log10(Total Sales + 1)", fontsize=12)
axes[1].set_ylabel("Number of Game-Console Releases", fontsize=12)
axes[1].tick_params(axis="both", labelsize=10)

plt.subplots_adjust(wspace=0.25)
plt.show()

print('The raw `total_sales` distribution is strongly right-skewed. Most game-console releases have relatively low sales, while a small number of major releases sell millions of copies. This matches the summary statistics, where the median platform-level sales value was 180,000 copies but the maximum was over 20 million copies.')
```

### Experimental Analysis: Do these variables behave the same/different across genres?
### Standard ordinary least squares (OLS) linear regression

#### Comparing Action Games to Adventure Games

- Step 1: Load & prepare data (Filter by genre, handle nulls, split Action vs Adventure)
- Step 2: Feature engineering (Log sales, publisher rank, platform count, scores)
- Step 3: Train on Action games (Fit LinearRegression, check R², coefficients)
- Step 4: Predict Adventure sales (Apply Action model to Adventure features)
- Step 5: Evaluate accuracy (RMSE, MAE, R², predicted vs empirical plot)

```python
action = games[games['genre'] == 'Action'].copy()
adventure = games[games['genre'] == 'Adventure'].copy()

features = ['publisher_rank', 'platform_release_count',
            'metacritic_review_score', 'user_review_score']
target = 'log_total_sales_per_title'

X_action = action[features]
y_action = action[target]
model = LinearRegression()
model.fit(X_action, y_action)

# Inspect model fit on training data (Action genre)
y_action_pred = model.predict(X_action)
print("=== Model fit on Action games ===")
print(f"R²:   {r2_score(y_action, y_action_pred):.3f}")
print(f"RMSE: {np.sqrt(mean_squared_error(y_action, y_action_pred)):.3f}")
print("\nCoefficients:")
for feat, coef in zip(features, model.coef_):
    print(f"  {feat}: {coef:.4f}")
print(f"  Intercept: {model.intercept_:.4f}")

# Apply the model to Adventure games
X_adventure = adventure[features]
y_adventure_empirical = adventure[target]

# Predict using the Action-trained model (no retraining)
y_adventure_predicted = model.predict(X_adventure)

# Convert back from log scale to raw sales for interpretability
adventure['Predicted_Sales'] = np.expm1(y_adventure_predicted)
adventure['Empirical_Sales'] = np.expm1(y_adventure_empirical)

# Compare predicted vs empirical & evaluate accuracy
print("")
print("=== Cross-genre accuracy (Action → Adventure) ===")
r2  = r2_score(y_adventure_empirical, y_adventure_predicted)
rmse = np.sqrt(mean_squared_error(y_adventure_empirical, y_adventure_predicted))
mae  = mean_absolute_error(y_adventure_empirical, y_adventure_predicted)

print(f"R²:   {r2:.3f}   (1.0 = perfect, 0 = no explanatory power)")
print(f"RMSE: {rmse:.3f}  (in log-sales units)")
print(f"MAE:  {mae:.3f}   (in log-sales units)")

adventure_model = LinearRegression()
adventure_model.fit(X_adventure, y_adventure_empirical)

coef_df = pd.DataFrame({
    'action':    model.coef_,
    'adventure': adventure_model.coef_
}, index=features)

coef_df['absolute_change']   = (coef_df['adventure'] - coef_df['action']).abs()
coef_df['relative_change_%'] = (
    (coef_df['adventure'] - coef_df['action']) / coef_df['action'].abs() * 100
).round(1)
coef_df['direction_flipped'] = (coef_df['action'] * coef_df['adventure']) < 0

print("\n=== Coefficient comparison: Action vs Adventure ===")
print(coef_df.to_string())

# ── RETRAIN ON ADVENTURE ──────────────────────────────────────────────────────

adventure_model = LinearRegression()
adventure_model.fit(X_adventure, y_adventure_empirical)
y_adventure_native = adventure_model.predict(X_adventure)

print("=== Model fit on Adventure games (retrained) ===")
print(f"R²:   {r2_score(y_adventure_empirical, y_adventure_native):.3f}")
print(f"RMSE: {np.sqrt(mean_squared_error(y_adventure_empirical, y_adventure_native)):.3f}")
print("\nCoefficients:")
for feat, coef in zip(features, adventure_model.coef_):
    print(f"  {feat}: {coef:.4f}")
print(f"  Intercept: {adventure_model.intercept_:.4f}")

# ── COMPARE ACTION vs ADVENTURE MODEL FIT ────────────────────────────────────
print("\n=== Retrained Model comparison ===")
print(f"{'':28} {'Action model':>14} {'Adventure model':>16}")
print("-" * 60)
print(f"{'R² (in-genre)':28} {r2_score(y_action, y_action_pred):>14.3f} {r2_score(y_adventure_empirical, y_adventure_native):>16.3f}")
print(f"{'RMSE (in-genre)':28} {np.sqrt(mean_squared_error(y_action, y_action_pred)):>14.3f} {np.sqrt(mean_squared_error(y_adventure_empirical, y_adventure_native)):>16.3f}")
print(f"{'R² on Adventure (transfer)':28} {r2_score(y_adventure_empirical, y_adventure_predicted):>14.3f} {'n/a':>16}")
```

#### Interpretation
- The relationships partially transfer. The model isn't useless on Adventure data, but it loses meaningful explanatory power.
- Adventure sales have dynamics the Action model doesn't fully capture, possibly different genre-specific audience size, platform preferences, or publisher strategies.

#### Relationships that DID NOT Change significantly Cross-Genre
- publisher_rank (−19.3%): bigger publishers sell more in both genres.
- platform_release_count(+55.5%): more platforms means more sales but the effect is slightly stronger for Adventure games (0.229 → 0.356).

#### Relationships that DID Change significantly Cross-Genre
- metacritic_review_score (−106.4%): Critic scores are positively associated with Action sales (+0.164) but negatively associated with Adventure sales (−0.011).
- user_review_score (+601.9%): User scores go from slightly negative in Action (−0.030) to meaningfully positive in Adventure (+0.152).

#### Evidence by the Retrained Model
The critic/user score flips confirmed earlier are now locked in by the retrained model, reinforcing that these aren't artifacts of the transfer. When we retrained the model purely on Adventure data, the model learned from scratch what predicts Adventure sales, and it still came back with a negative coefficient for critic scores (−0.0105) and a positive one for user scores (+0.1521).

#### The Bottom Line
Higher critic scores positively predict sale for action games but is slightly negative for adventure games. On the other hand, higher user scores positively predict sales for adventure games but is slightly negative for action games. In conclusion, adventure game enjoyers are a community-driven audience that buys based on peer reception rather than critical consensus.
